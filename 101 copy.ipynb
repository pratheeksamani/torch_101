```python
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, Dataset
import torch.optim as optim
```

```python
class CustomDataset(Dataset):
    def __init__(self, data_size):
        self.X = torch.randn(data_size, 4, dtype=torch.float32)
        self.logits = (self.X[:, 0] + self.X[:, 1] - self.X[:, 2] + 0.5 * self.X[:, 3])
        self.y = (self.logits > 0).float().unsqueeze(1)
    def __len__(self)-> int:
        return len(self.X)

    def __getitem__(self, idx)-> tuple[torch.Tensor, torch.Tensor]:
        return self.X[idx], self.y[idx]
```

```python
class customModel(nn.Module):
    def __init__(self, input_size=4, hidden_size=16, hidden_size2=16, output_size=1):
        super(customModel, self).__init__()
        self.fc1 = nn.Linear(input_size, hidden_size)
        self.fc2 = nn.Linear(hidden_size, hidden_size2)
        self.fc3 = nn.Linear(hidden_size2, output_size)
        self.Relu = nn.ReLU()

    def forward(self, x) -> torch.Tensor:
        x = self.Relu(self.fc1(x))
        x = self.Relu(self.fc2(x))
        x = self.fc3(x)
        return x
```

```python
def pipeline():
    # Hyperparameters
    data_size = 1000
    batch_size = 32
    learning_rate = 0.001
    num_epochs = 20

    # Dataset and DataLoader
    dataset = CustomDataset(data_size)
    dataloader = DataLoader(dataset, batch_size=batch_size, shuffle=True)

    # Model, Loss, Optimizer
    model = customModel(input_size = 4)
    criterion = nn.BCEWithLogitsLoss()
    optimizer = optim.Adam(model.parameters(), lr=learning_rate)

    # Training Loop
    for epoch in range(num_epochs):
        model.train()
        epoch_loss = 0.0
        batch_count = 0
        for inputs, labels in dataloader:
            optimizer.zero_grad()
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()
            epoch_loss += loss.item()
            batch_count += 1
        avg_loss = epoch_loss / batch_count
        print(f'Epoch [{epoch+1}/{num_epochs}], Loss: {avg_loss:.4f}')
    return model
```

```python
if __name__ == "__main__":
    model = pipeline()
```
